## 2. Marco teórico: Adaline (Adaptive Linear Element)

Algoritmo de aprendizaje supervisado para la clasificación binaria y la regresión lineal.
Tanto Adaline como Perceptron son modelos de redes neuronales (de una sola capa).

Utiliza la regla de aprendizaje de Widrow-Hoff (least-mean squares).
ADALINE utiliza una función de activación lineal y el descenso de gradiente para ajustar 
los pesos sinápticos y encontrar la mejor solución posible.

**Algoritmo Logicay metodologia :**
1. Inicialización: Se inicializan los pesos sinápticos aleatoriamente y se establece una tasa de aprendizaje (learning rate).
2. Propagación hacia adelante: Se aplica la función de activación lineal a los datos ponderados. El resultado se compara con la salida deseada.
3. Cálculo del error: Diferencia entre la salida deseada y la salida real.
4. Actualización de pesos: Se ajustan los pesos utilizando el descenso de gradiente para minimizar el error.
5. Repetición: Se repiten los pasos 2-4 para todo el conjunto de entrenamiento.
6. Evaluación: Se evalúa el modelo en un conjunto de prueba.

# 3. Práctica 1: ADALINE con Gradiente Descendente

Se ajusta un polinomio de grado 4 a la función conocida $Y = 15 - 3x + 4x^2 - 6x^4$ usando una red ADALINE entrenada con gradiente descendente (GD), y se compara contra la solución exacta por mínimos cuadrados (MC).

<a id="p1-definicion"></a>

### 3.1 Definición del problema

En esta práctica se busca realizar una aproximación polinomial utilizando el método de mínimos cuadrados. Para ello, se generan 100 datos aleatorios y se calculan sus valores mediante una función conocida. Posteriormente, se ajustan polinomios de grado 1 hasta grado 4, con el propósito de comparar qué tan bien representan los datos originales. A través de la función de costo y las gráficas obtenidas, se busca identificar qué grado de polinomio proporciona el mejor ajuste.

<a id="p1-algoritmo"></a>

### 3.2 Descripción del algoritmo

El algoritmo utiliza el método de mínimos cuadrados para calcular los coeficientes de diferentes polinomios. Primero, se generan los datos y se establece el grado máximo de aproximación. Después, para cada grado, se construye una matriz con las potencias de los valores de entrada y se calculan los pesos que permiten obtener la aproximación. Finalmente, se determina el error, se calcula la función de costo y se representan gráficamente los resultados para comparar los modelos.

**Pseudocódigo**
```
INICIO
    Generar 100 valores aleatorios de X
    Calcular Y = 15 - 3X + 4X² - 6X⁴
    Obtener el número de datos m
    Inicializar Xa con una columna de unos
    Definir grado máximo = 4
    Inicializar vector de costos J

    PARA i = 1 HASTA 4 HACER
        Agregar X^i a la matriz Xa
        Calcular pesos W mediante mínimos cuadrados
        Calcular salida estimada Yg = Xa * W
        Calcular error E = Y - Yg
        Calcular costo J = (Eᵀ * E) / (2*m)
        Graficar valores reales y estimados
    FIN PARA

    Mostrar gráficas de cada grado
    Graficar función de costo por grado
    Mostrar pesos del último polinomio
FIN
```

### 3.3 Implementación

**3.3.1 Exploración inicial por mínimos cuadrados (grados 1 a 4)**

````python


# Limpieza de entorno manejada nativamente por Python
# Esta sección solo es de prueba debe ser borrada, debido a que trataremos 
# de ajustar un polinomio a una función "conocida" solo para medir el 
# rendimiento del algoritmo.

# Fijamos una semilla para reproducibilidad en las pruebas (opcional)
#np.random.seed(42)

X = np.random.rand(100, 1) # Se generan 100 números aleatorios
Y = 15 - 3*X + 4*(X**2) - 6*(X**4) # Función conocida

# SECCIÓN POR MÍNIMOS CUADRADOS 
# (Este paso es solo para verificar los coeficientes de aproximación)
#
# 1) Modelo: Y_hat = Xa * W
# 2) Error: E = Y - Y_hat
# 3) Función de costo: J = (1/2m) * E^T * E
# 4) Pesos: W = (Xa^T * Xa)^-1 * Xa^T * Y

m = X.shape[0] # Tamaño de los datos
Xa = np.ones((m, 1)) # Matriz de 1 y valores de x
grado = 4 # Grado del polinomio de aproximación

J_mc = np.zeros(grado) # con puros ceros porque va ir calculando cada aproximacion por indivisual 

# Configuración de subgráficas (equivalente a nexttile) va a tener 4 subgraficas 
fig, axs = plt.subplots(2, 2, figsize=(10, 8))
axs = axs.flatten()

# Crear el modelo de 1 hasta el grado del polinomio
for i in range(1, grado + 1):
    # Matriz con valores de Xa ejemplo Xa=[1 X X^2 X^3 X^4]
    Xa = np.hstack((Xa, X**i)) 
    
    # Calcular los pesos W_mc
    W_mc = np.linalg.inv(Xa.T @ Xa) @ Xa.T @ Y 
    
    # Calcular la salida del modelo
    Yg_mc = Xa @ W_mc 
    
    # Calcular el error E
    E = Y - Yg_mc 
    
    # Obtener la funcion de costo J
    # Extraemos el valor escalar con [0, 0] ya que el producto matricial devuelve matriz 1x1
    J_mc[i-1] = (E.T @ E)[0, 0] / (2 * m) 
    
    # Graficar X, Yg_mc y X, Y
    axs[i-1].plot(X, Yg_mc, "o", label='Modelo')
    axs[i-1].plot(X, Y, "+", label='Reales')
    axs[i-1].set_title(f'Grado {i}')
    axs[i-1].legend()

plt.tight_layout()
plt.show()

# Graficar la evolución de la función de costo por grado
plt.figure()
plt.plot(range(1, grado + 1), J_mc, 'b-o') #
plt.xlabel('Grado del polinomio') #
plt.ylabel('J(X,W)') #
plt.title('Costo por Mínimos Cuadrados')
plt.grid(True)
plt.show()

print("Pesos por mínimos cuadrados (W_mc):")
print(W_mc)
````

**3.3.2 ADALINE con Gradiente Descendente**

````python
# ------------------------------------------------------------
# 1. LIBRERIAS Y GENERACION DE LOS DATOS
# ------------------------------------------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Semilla para obtener los mismos datos en cada ejecucion
np.random.seed(42)

# 100 valores aleatorios entre 0 y 1
X = np.random.rand(100, 1)

# Funcion que queremos aproximar
Y = 15 - 3*X + 4*(X**2) - 6*(X**4)

m = X.shape[0]     # Numero de observaciones
grado = 4          # Grado del polinomio
````

````python
# ------------------------------------------------------------
# 2. MATRIZ DE CARACTERISTICAS Y NORMALIZACION
# ------------------------------------------------------------
def construir_Xa(X, grado):
    # Columna de unos (sesgo) + potencias de X hasta el grado indicado
    Xa = np.ones((X.shape[0], 1))
    for i in range(1, grado + 1):
        Xa = np.hstack((Xa, X**i))
    return Xa

def normalizar(Xa, min_vals, max_vals):
    # Normalizacion min-max de las columnas de potencias (no del sesgo)
    Xn = Xa.copy()
    Xn[:, 1:] = (Xa[:, 1:] - min_vals) / (max_vals - min_vals)
    return Xn

def desnormalizar(W, min_vals, max_vals):
    # Convierte los pesos entrenados con datos normalizados a los
    # coeficientes ORIGINALES del polinomio (para poder compararlos con [15,-3,4,0,-6])
    rango = max_vals - min_vals
    w = W[1:, 0] / rango
    b = W[0, 0] - np.sum(W[1:, 0] * min_vals / rango)
    return np.concatenate(([b], w)).reshape(-1, 1)

Xa_raw = construir_Xa(X, grado)
min_vals = Xa_raw[:, 1:].min(axis=0)
max_vals = Xa_raw[:, 1:].max(axis=0)
Xa = normalizar(Xa_raw, min_vals, max_vals)
````

````python
# ------------------------------------------------------------
# 3. FUNCION DE ENTRENAMIENTO ADALINE (GRADIENTE DESCENDENTE)
# ------------------------------------------------------------
def adaline_gd(Xa, Y, eta, epocas):
    # Pesos iniciales en cero para que todas las pruebas partan igual
    W = np.zeros((Xa.shape[1], 1))
    J_historial = np.full(epocas, np.nan)
    n = len(Y)

    with np.errstate(all="ignore"):          # evita avisos por overflow si diverge
        for k in range(epocas):
            Y_estimado = Xa @ W                          # salida lineal
            E = Y - Y_estimado                           # error
            J_historial[k] = (E.T @ E)[0, 0] / (2 * n)   # costo
            if not np.isfinite(J_historial[k]):          # divergencia: detener
                break
            gradiente = -(Xa.T @ E) / n                  # gradiente
            W = W - eta * gradiente                      # actualizacion de pesos

        Y_estimado = Xa @ W
        E_final = Y - Y_estimado
        J_final = (E_final.T @ E_final)[0, 0] / (2 * n)
        RMSE = np.sqrt(np.mean(E_final**2))

    return W, J_historial, Y_estimado, J_final, RMSE
````

````python
# ------------------------------------------------------------
# 4. SOLUCION DE REFERENCIA: MINIMOS CUADRADOS
# ------------------------------------------------------------
# W = (Xa^T Xa)^-1 Xa^T Y  sobre la misma matriz normalizada
W_mc = np.linalg.inv(Xa.T @ Xa) @ Xa.T @ Y
E_mc = Y - Xa @ W_mc
J_mc = (E_mc.T @ E_mc)[0, 0] / (2 * m)
print(f"Costo minimos cuadrados: {J_mc:.3e}")
````

**3.3.3 Entrenamiento base**
Se usa $\eta = 1.0$ y 100 000 épocas (configuración elegida después de la tabla comparativa de la sección 6; con 5 000 épocas el modelo aún no converge).

````python
# ------------------------------------------------------------
# 5. ENTRENAMIENTO DEL MODELO Y GRAFICAS
# ------------------------------------------------------------
eta = 1.0
epocas = 100000

W_gd, J_gd, Y_gd, J_final, RMSE = adaline_gd(Xa, Y, eta, epocas)

print(f"Costo final: {J_final:.3e}")
print(f"RMSE final : {RMSE:.3e}")

# Pesos en la escala ORIGINAL (des-normalizados)
tabla_pesos = pd.DataFrame({
    "Coef. verdadero": [15, -3, 4, 0, -6],
    "Minimos cuadrados": desnormalizar(W_mc, min_vals, max_vals).ravel(),
    "ADALINE (GD)": desnormalizar(W_gd, min_vals, max_vals).ravel(),
}, index=["W0 (1)", "W1 (x)", "W2 (x^2)", "W3 (x^3)", "W4 (x^4)"])
tabla_pesos.round(5)
````

````python
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# Comparacion entre datos reales y estimados (ordenados por X para ver la curva)
idx = np.argsort(X[:, 0])
axs[0].scatter(X[:, 0], Y[:, 0], s=15, label="Datos reales")
axs[0].plot(X[idx, 0], Y_gd[idx, 0], "r-", lw=2, label="ADALINE - Gradiente Descendente")
axs[0].set(xlabel="X", ylabel="Y", title="Comparacion entre valores reales y estimados")
axs[0].legend(); axs[0].grid(True)

# Convergencia: costo por epoca (eje Y logaritmico para ver todo el descenso)
axs[1].semilogy(range(1, epocas + 1), J_gd)
axs[1].set(xlabel="Epocas", ylabel="Costo J (escala log)", title="Convergencia de ADALINE")
axs[1].grid(True, which="both", alpha=0.3)

plt.tight_layout(); plt.show()
````

### 3.4 Pruebas y resultados

**3.4.1 Tabla comparativa: tasa de aprendizaje vs. épocas**
Se amplía el rango de η (incluyendo valores muy pequeños y valores que provocan divergencia) y el de épocas, para observar los tres comportamientos posibles: convergencia lenta, convergencia rápida y divergencia.

````python
# ------------------------------------------------------------
# 6. PRUEBAS CON DIFERENTES TASAS DE APRENDIZAJE Y EPOCAS
# ------------------------------------------------------------
tasas = [0.001, 0.01, 0.05, 0.10, 0.5, 1.0, 1.5]
lista_epocas = [1000, 10000, 100000]

def estado(costo, J_prueba):
    if (not np.isfinite(costo)) or costo > J_prueba[0]:
        return "Diverge"
    if costo > 1e-3:  return "Lento"
    if costo > 1e-6:  return "Aceptable"
    return "Convergio"

resultados = []
for tasa in tasas:
    for num_epocas in lista_epocas:
        W_p, J_p, Y_p, costo, rmse = adaline_gd(Xa, Y, tasa, num_epocas)
        diverge = (not np.isfinite(costo)) or costo > J_p[0]
        resultados.append({
            "Tasa de aprendizaje": tasa,
            "Epocas": num_epocas,
            "Costo final": np.nan if diverge else costo,
            "RMSE": np.nan if diverge else rmse,
            "Dist. a pesos MC": np.nan if diverge else np.linalg.norm(W_p - W_mc),
            "Estado": estado(costo, J_p),
        })

tabla_resultados = pd.DataFrame(resultados)
print("TABLA COMPARATIVA DE ADALINE\n")
tabla_resultados
````

**3.4.2 Validación, métricas y análisis de residuos**
Se divide el conjunto en 80 % entrenamiento y 20 % prueba. La normalización se calcula **solo con los datos de entrenamiento** (para no filtrar información del conjunto de prueba) y se aplica después a ambos.

````python
# ------------------------------------------------------------
# 8. VALIDACION, METRICAS Y RESIDUOS
# ------------------------------------------------------------
# 1. Division 80/20 (sin sklearn, con numpy)
rng = np.random.default_rng(42)
perm = rng.permutation(m)
n_train = int(0.8 * m)
idx_train, idx_test = perm[:n_train], perm[n_train:]

Xr_train, Xr_test = Xa_raw[idx_train], Xa_raw[idx_test]
Y_train, Y_test = Y[idx_train], Y[idx_test]

# 2. Normalizacion con estadisticas SOLO del entrenamiento
mn_tr = Xr_train[:, 1:].min(axis=0)
mx_tr = Xr_train[:, 1:].max(axis=0)
Xa_train = normalizar(Xr_train, mn_tr, mx_tr)
Xa_test  = normalizar(Xr_test,  mn_tr, mx_tr)

# 3. Reentrenamiento con la mejor configuracion de la tabla
eta_optimo, epocas_optimas = 1.0, 100000
W_train, J_hist_train, Y_est_train, J_final_train, RMSE_train = adaline_gd(
    Xa_train, Y_train, eta_optimo, epocas_optimas)

# 4. Evaluacion en datos no vistos
Y_est_test = Xa_test @ W_train
E_test = Y_test - Y_est_test
RMSE_test = np.sqrt(np.mean(E_test**2))
R2_test = 1 - np.sum(E_test**2) / np.sum((Y_test - Y_test.mean())**2)

print("--- EVALUACION EN CONJUNTO DE PRUEBA ---")
print(f"RMSE en Entrenamiento: {RMSE_train:.3e}")
print(f"RMSE en Prueba       : {RMSE_test:.3e}")
print(f"R^2 en Prueba        : {R2_test:.8f}")
print("(Un R^2 cercano a 1.0 indica un ajuste excelente)\n")

# 5. Pesos en escala original (comparables con [15, -3, 4, 0, -6])
W_orig = desnormalizar(W_train, mn_tr, mx_tr)
print("--- PESOS FINALES (escala original) ---")
for nombre, w in zip(["Bias (W0)", "X (W1)", "X^2 (W2)", "X^3 (W3)", "X^4 (W4)"], W_orig[:, 0]):
    print(f"{nombre:10s}: {w:.5f}")
````

**3.4.3 Análisis de resultados**

**1. Efecto de la tasa de aprendizaje (η).**
Con η pequeña el costo baja muy despacio: con η = 0.001 y 100 000 épocas el costo sigue en ≈ 9.9×10⁻³ (RMSE ≈ 0.14). Al aumentar η el descenso por época es mucho más rápido: con η = 1.0 se llega a J ≈ 1.5×10⁻⁵ en 10 000 épocas y a J ≈ 6.5×10⁻⁸ (RMSE ≈ 3.6×10⁻⁴) en 100 000. Los mejores resultados de la tabla son η = 0.5 y η = 1.0 con 100 000 épocas.

**2. Efecto del número de épocas.**
Para una misma η, más épocas siempre reducen el costo, pero con rendimientos distintos: con η = 0.1, pasar de 1 000 a 10 000 épocas apenas reduce J de 9.9×10⁻³ a 5.0×10⁻³, y es hasta las 100 000 épocas cuando baja a 1.5×10⁻⁵. En la gráfica de convergencia se observa que todas las curvas tienen una **meseta** alrededor de J ≈ 10⁻² y después caen bruscamente; una η mayor adelanta esa caída.

**3. Relación entre η y épocas.**
Lo que importa es aproximadamente el producto η × épocas: (η = 0.001, 10 000 épocas) y (η = 0.01, 1 000 épocas) dan casi el mismo costo (≈ 0.159); igual ocurre con (0.01, 10 000) y (0.1, 1 000), ambos ≈ 9.93×10⁻³. Subir η equivale a dar pasos más grandes y permite ahorrar épocas.

**4. Divergencia.**
Con η = 1.5 el costo crece sin control en las tres duraciones probadas (el entrenamiento se detiene al detectar `inf`/`nan`), por lo que se marca como *Diverge*. Existe, por tanto, un límite superior de η entre 1.0 y 1.5 (para este problema, 2/λ_max ≈ 1.31). Se recomienda elegir una η cercana pero por debajo de ese límite.

**5. Por qué la convergencia es lenta.**
Las columnas x, x², x³ y x⁴ están muy correlacionadas entre sí (aunque estén normalizadas), lo que forma un “valle” alargado en la superficie de costo: el GD avanza rápido en una dirección y muy lento en la otra. De ahí la meseta y la necesidad de cientos de miles de épocas. Una normalización z o centrar los datos mejoraría la velocidad.

**6. Costo bajo no implica pesos iguales a los reales.**
Con η = 1.0 y 100 000 épocas el ajuste es excelente (RMSE ≈ 3.6×10⁻⁴), pero los pesos todavía difieren de los reales: W₃ ≈ 0.157 en vez de 0 y W₂ ≈ 3.90 en vez de 4. Al haber variables colineales, distintas combinaciones de coeficientes producen casi la misma curva. Los pesos solo se acercan por completo a los de mínimos cuadrados con más épocas.

**7. Validación.**
En el conjunto de prueba (20 % de los datos, no vistos) se obtiene RMSE ≈ 3.7×10⁻⁴, prácticamente igual al de entrenamiento (3.7×10⁻⁴), y R² ≈ 0.9999999: **no hay sobreajuste** y el modelo generaliza. Cabe señalar que los datos no tienen ruido, por lo que un R² alto era esperable; la métrica útil para comparar configuraciones es el RMSE/costo, no el R². Los residuos son del orden de 10⁻³ y no muestran un patrón claro.

**8. Comparación con mínimos cuadrados.**
Mínimos cuadrados obtiene la solución exacta en un solo cálculo (J ≈ 10⁻²³, ruido numérico), mientras que el GD solo la aproxima de forma iterativa. La ventaja del GD es que no requiere invertir XᵀX, y es la base del entrenamiento en redes neuronales más grandes.

**Conclusión:** la configuración más adecuada fue **η = 1.0 con 100 000 épocas** (η = 0.5 es una alternativa más conservadora). Valores η ≥ 1.5 divergen y valores η ≤ 0.01 requieren un número de épocas impracticable.

# 4. Práctica 2: Regresión con ADALINE

<a id="p2-definicion"></a>

### 4.1 Definición del problema

En la Práctica 1 la función a aproximar era **conocida**, por lo que se podía comparar directamente con los coeficientes reales. En la Práctica 2 el problema es más realista: se tiene un conjunto de datos $\{(x^{(k)}, y^{(k)})\}_{k=1}^{m}$ en el archivo *"Regression Task.xlsx"* y **no se conoce la función que los generó**. El objetivo es encontrar un modelo que prediga la variable de salida $y$ a partir de las variables de entrada.

Se proponen tres problemas de regresión, uno por cada hoja del archivo:

| Hoja | Variables de entrada | Salida | Observaciones |
|---|---|---|---|
| Problem 1 | $x$ | $y$ | 200 |
| Problem 2 | $x_1, x_2$ | $y$ | 300 |
| Problem 3 | $x_1, x_2, x_3$ | $y$ | 200 |

Como la relación entre entradas y salida puede ser no lineal, se usa una **regresión polinomial**: el modelo es lineal en los pesos pero no en las variables. Para una variable y grado $g$:

$$\hat{y} = w_0 + w_1 x + w_2 x^2 + \dots + w_g x^g = X_a W$$

Esto permite seguir usando una red ADALINE (que solo sabe hacer combinaciones lineales) si se le dan como entradas las potencias $x, x^2, \dots, x^g$. El problema, entonces, tiene dos partes:

1. **Elegir el grado $g$** del polinomio. Un grado muy bajo no captura la forma de los datos (*subajuste*) y un grado muy alto agrega términos innecesarios que pueden ajustar ruido (*sobreajuste*) y hacen el entrenamiento más lento e inestable. Para elegirlo se usa el **método del codo de Jambú**.
2. **Encontrar los pesos $W$** que minimizan la función de costo
$$J(W) = \frac{1}{2m}\,(Y - X_a W)^T (Y - X_a W)$$
mediante el entrenamiento de ADALINE por gradiente descendente, verificando el resultado con la solución analítica de mínimos cuadrados.

<a id="p2-algoritmo"></a>

### 4.2 Descripción del algoritmo

El algoritmo de la Práctica 1 se adapta a la regresión con datos reales agregando tres etapas: la **preparación de los datos**, la **selección del grado** con el codo de Jambú y la **validación** con datos no vistos.

**1. Carga y limpieza de datos.** Se leen las columnas de la hoja de Excel, se eliminan columnas vacías o con celdas basura, filas con valores faltantes o no numéricos y registros duplicados.

**2. Construcción de la matriz de características.** Para un grado $g$ se forma $X_a = [\,1,\ x,\ x^2,\ \dots,\ x^g\,]$. En los problemas de 2 y 3 variables se agregan también los productos cruzados (por ejemplo $x_1 x_2$, $x_1^2 x_2$).

**3. Normalización z.** Las potencias de $x$ tienen escalas muy distintas y están muy correlacionadas, lo que hace lento al gradiente descendente (como se observó en la Práctica 1 con la normalización min-max). Por eso cada columna (excepto la de unos) se estandariza:
$$z_j = \frac{x_j - \mu_j}{\sigma_j}$$
Las medias $\mu_j$ y desviaciones $\sigma_j$ se calculan **solo con los datos de entrenamiento**.

**4. Codo de Jambú.** Para cada grado $g = 1, 2, \dots, g_{max}$ se entrena el modelo y se guarda su costo $J(g)$. Al graficar $J$ contra $g$, el costo baja rápidamente mientras el polinomio gana capacidad para representar los datos, y después se aplana porque los términos extra ya no aportan. El **grado óptimo es el punto donde la curva "dobla"** (el codo): el menor grado a partir del cual aumentar el grado ya no reduce el costo de forma significativa.

**5. Entrenamiento ADALINE (gradiente descendente)** con el grado elegido:
- Salida: $\hat{Y} = X_a W$
- Error: $E = Y - \hat{Y}$
- Gradiente: $\nabla J = -\frac{1}{m} X_a^T E$
- Actualización: $W \leftarrow W - \eta \nabla J$

La tasa de aprendizaje debe cumplir $\eta < 2/\lambda_{max}$, donde $\lambda_{max}$ es el mayor eigenvalor de $\frac{1}{m}X_a^T X_a$; de lo contrario el entrenamiento diverge.

**6. Validación y desnormalización.** Se evalúa el modelo en el conjunto de prueba (RMSE y $R^2$) y los pesos se regresan a la escala original para interpretarlos como coeficientes del polinomio:
$$w_j^{orig} = \frac{w_j}{\sigma_j}, \qquad w_0^{orig} = w_0 - \sum_j \frac{w_j\,\mu_j}{\sigma_j}$$

**Pseudocódigo**
```
INICIO
    Leer hoja de Excel -> X, Y
    Limpiar datos (columnas vacías, NaN, no numéricos, duplicados)
    Dividir en entrenamiento (80 %) y prueba (20 %)

    // ---- Codo de Jambú ----
    PARA g = 1 HASTA g_max HACER
        Xa <- [1, x, x^2, ..., x^g]       (normalizado con media/desv. de entrenamiento)
        W  <- (Xa^T Xa)^-1 Xa^T Y         (mínimos cuadrados, referencia exacta)
        J_mc(g) <- costo de entrenamiento y de prueba
        W  <- ADALINE_GD(Xa, Y, eta, epocas)
        J_gd(g) <- costo de ADALINE
    FIN PARA
    Graficar J vs g y elegir g* en el codo

    // ---- Modelo final ----
    Xa <- [1, x, ..., x^g*] normalizado
    W  <- 0
    PARA k = 1 HASTA epocas HACER
        Y_est <- Xa W
        E     <- Y - Y_est
        J(k)  <- (E^T E) / (2m)
        W     <- W + eta * Xa^T E / m
    FIN PARA
    Evaluar RMSE y R^2 en prueba
    Desnormalizar W y comparar con mínimos cuadrados
FIN
```

### 4.3 Problem 1: regresión de 1 variable

<a id="p2-p1-implementacion"></a>

#### 4.3.1 Implementación

**a) Carga de datos.** Se lee la hoja *"Problem 1"*. Se lee sin restringir columnas para poder revisar si el archivo trae basura además de las columnas `x` y `y`.

````python
# ------------------------------------------------------------
# PRÁCTICA 2 - PROBLEM 1: CARGA DE DATOS
# ------------------------------------------------------------
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Ruta del archivo: se busca primero junto al notebook y luego en la carpeta del Módulo 2
rutas = ["Regression Task.xlsx", os.path.join("..", "Clases", "Modulo_2", "Regression Task.xlsx")]
archivo = next(r for r in rutas if os.path.exists(r))

df_crudo = pd.read_excel(archivo, sheet_name="Problem 1")

print(f"Dimensiones originales: {df_crudo.shape}")
print(f"Columnas: {df_crudo.columns.tolist()}\n")
df_crudo.head()
````

**b) Limpieza de datos.** Se revisan valores faltantes, columnas extra, datos no numéricos y duplicados.

````python
# ------------------------------------------------------------
# LIMPIEZA DE DATOS
# ------------------------------------------------------------
# 1. Diagnóstico: valores no vacíos por columna
print("Valores no vacíos por columna:")
print(df_crudo.notna().sum(), "\n")

# 2. Las columnas 'Unnamed' son columnas vacías del Excel (una trae un espacio en blanco);
#    solo se conservan 'x' y 'y'
df = df_crudo[["x", "y"]].copy()

# 3. Forzar a numérico: cualquier texto se convierte en NaN
df = df.apply(pd.to_numeric, errors="coerce")

# 4. Eliminar filas incompletas y registros duplicados
n_antes = len(df)
df = df.dropna().drop_duplicates().reset_index(drop=True)
print(f"Filas eliminadas por NaN o duplicados: {n_antes - len(df)}")
print(f"Dimensiones después de la limpieza: {df.shape}\n")

# 5. Estadística descriptiva para revisar rangos y posibles valores atípicos
df.describe()
````

````python
# Vectores columna (m x 1), igual que en la Práctica 1
X_p1 = df["x"].to_numpy().reshape(-1, 1)
Y_p1 = df["y"].to_numpy().reshape(-1, 1)
m_p1 = X_p1.shape[0]

# Gráfica de dispersión para ver la forma de los datos
plt.figure(figsize=(7, 4.5))
plt.scatter(X_p1, Y_p1, s=15)
plt.xlabel("x"); plt.ylabel("y")
plt.title("Problem 1: datos originales")
plt.grid(True)
plt.show()
````

La nube de puntos muestra una relación **no lineal y decreciente** (curva cóncava hacia abajo), por lo que una recta no será suficiente y se justifica probar polinomios de mayor grado.

**c) Funciones del modelo.** Se reutiliza la misma estructura de la Práctica 1, cambiando la normalización min-max por normalización z.

````python
# ------------------------------------------------------------
# FUNCIONES: CARACTERÍSTICAS, NORMALIZACIÓN Y ENTRENAMIENTO
# ------------------------------------------------------------
def potencias(X, grado):
    # Columnas [x, x^2, ..., x^grado] (sin la columna de unos)
    return np.hstack([X**i for i in range(1, grado + 1)])

def matriz_Xa(X, grado, mu, sigma):
    # Columna de unos (sesgo) + potencias estandarizadas con normalización z
    Z = (potencias(X, grado) - mu) / sigma
    return np.hstack((np.ones((X.shape[0], 1)), Z))

def costo(Xa, Y, W):
    # J = (1/2m) E^T E
    with np.errstate(all="ignore"):         # evita avisos si los pesos divergieron
        E = Y - Xa @ W
        return (E.T @ E)[0, 0] / (2 * len(Y))

def minimos_cuadrados(Xa, Y):
    # W = (Xa^T Xa)^-1 Xa^T Y  (se usa solve en lugar de inv por estabilidad numérica)
    return np.linalg.solve(Xa.T @ Xa, Xa.T @ Y)

def adaline_gd(Xa, Y, eta, epocas):
    # Entrenamiento ADALINE por gradiente descendente (regla de Widrow-Hoff en lote)
    W = np.zeros((Xa.shape[1], 1))          # pesos iniciales en cero
    J_hist = np.full(epocas, np.nan)
    m = len(Y)
    with np.errstate(all="ignore"):
        for k in range(epocas):
            E = Y - Xa @ W                       # error
            J_hist[k] = (E.T @ E)[0, 0] / (2 * m)
            if not np.isfinite(J_hist[k]):       # se detiene si diverge
                break
            W = W + eta * (Xa.T @ E) / m         # W <- W - eta * gradiente
    return W, J_hist

def desnormalizar_z(W, mu, sigma):
    # Convierte los pesos entrenados con datos estandarizados a coeficientes del polinomio original
    w = W[1:, 0] / sigma
    b = W[0, 0] - np.sum(W[1:, 0] * mu / sigma)
    return np.concatenate(([b], w))

def eta_maxima(Xa):
    # Límite de estabilidad del gradiente descendente: eta < 2 / lambda_max
    return 2 / np.linalg.eigvalsh(Xa.T @ Xa / Xa.shape[0]).max()
````

**d) División entrenamiento / prueba.** Se separa el 80 % de los datos para entrenar y el 20 % para validar, con semilla fija para que los resultados sean reproducibles.

````python
rng = np.random.default_rng(42)
perm = rng.permutation(m_p1)
n_train = int(0.8 * m_p1)
idx_tr, idx_te = perm[:n_train], perm[n_train:]

X_tr, Y_tr = X_p1[idx_tr], Y_p1[idx_tr]
X_te, Y_te = X_p1[idx_te], Y_p1[idx_te]
print(f"Entrenamiento: {len(X_tr)} datos | Prueba: {len(X_te)} datos")
````

<a id="p2-p1-pruebas"></a>

#### 4.3.2 Pruebas y resultados: codo de Jambú

Se entrenan polinomios de grado 1 a 8. Para cada grado se calcula:
- el costo con **mínimos cuadrados** (mínimo exacto que puede alcanzar ese grado), en entrenamiento y en prueba;
- el costo alcanzado por **ADALINE** con gradiente descendente ($\eta$ fija y 20 000 épocas).

````python
# ------------------------------------------------------------
# CODO DE JAMBÚ: COSTO vs GRADO DEL POLINOMIO
# ------------------------------------------------------------
grado_max = 8
eta_jambu = 0.2            # menor que 2/lambda_max para todos los grados probados
epocas_jambu = 20000

filas = []
for g in range(1, grado_max + 1):
    # Normalización con estadísticas del entrenamiento
    P_tr = potencias(X_tr, g)
    mu, sigma = P_tr.mean(axis=0), P_tr.std(axis=0)
    Xa_tr = matriz_Xa(X_tr, g, mu, sigma)
    Xa_te = matriz_Xa(X_te, g, mu, sigma)

    # Referencia exacta: mínimos cuadrados
    W_mc = minimos_cuadrados(Xa_tr, Y_tr)

    # ADALINE con gradiente descendente
    W_gd, _ = adaline_gd(Xa_tr, Y_tr, eta_jambu, epocas_jambu)

    filas.append({
        "Grado": g,
        "J MC (entren.)": costo(Xa_tr, Y_tr, W_mc),
        "J MC (prueba)": costo(Xa_te, Y_te, W_mc),
        "J ADALINE (entren.)": costo(Xa_tr, Y_tr, W_gd),
        "J ADALINE (prueba)": costo(Xa_te, Y_te, W_gd),
        "eta máx.": eta_maxima(Xa_tr),
    })

tabla_jambu = pd.DataFrame(filas).set_index("Grado")
# Mejora al subir del grado anterior a este, como % del costo del modelo lineal (grado 1)
J_lin = tabla_jambu.loc[1, "J MC (entren.)"]
tabla_jambu["Mejora vs grado anterior (% de J grado 1)"] = -tabla_jambu["J MC (entren.)"].diff() / J_lin * 100
pd.set_option("display.float_format", "{:.3e}".format)
tabla_jambu
````

````python
# Gráficas del codo de Jambú
grados = tabla_jambu.index.to_numpy()
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# (a) Escala lineal: es la forma clásica de ver el codo
axs[0].plot(grados, tabla_jambu["J MC (entren.)"], "b-o", label="Mínimos cuadrados")
axs[0].plot(grados, tabla_jambu["J ADALINE (entren.)"], "r--s", mfc="none", label="ADALINE (GD)")
axs[0].annotate("Codo (grado 3)", xy=(3, tabla_jambu.loc[3, "J MC (entren.)"]),
                xytext=(4.2, 0.006), arrowprops=dict(arrowstyle="->"))
axs[0].set(xlabel="Grado del polinomio", ylabel="J(X,W)", title="Codo de Jambú (escala lineal)")
axs[0].legend(); axs[0].grid(True)

# (b) Escala logarítmica: permite ver los costos muy pequeños después del codo
axs[1].semilogy(grados, tabla_jambu["J MC (entren.)"].clip(lower=1e-32), "b-o", label="MC entrenamiento")
axs[1].semilogy(grados, tabla_jambu["J MC (prueba)"].clip(lower=1e-32), "g-^", label="MC prueba")
axs[1].semilogy(grados, tabla_jambu["J ADALINE (entren.)"], "r--s", mfc="none", label="ADALINE entrenamiento")
axs[1].axvline(3, color="gray", ls=":")
axs[1].set(xlabel="Grado del polinomio", ylabel="J(X,W) (escala log)", title="Codo de Jambú (escala logarítmica)")
axs[1].legend(); axs[1].grid(True, which="both", alpha=0.3)

plt.tight_layout(); plt.show()
````

````python
# Selección automática del codo: el menor grado tal que subir un grado más
# reduce el costo en menos del 1 % del costo del modelo lineal (grado 1)
J = tabla_jambu["J MC (entren.)"]
umbral = 0.01 * J.loc[1]
grado_opt = next(g for g in grados[:-1] if J.loc[g] - J.loc[g + 1] < umbral)
print(f"Grado óptimo según el codo de Jambú: {grado_opt}")
````

````python
# Comparación visual del ajuste para los grados 1 a 4
x_linea = np.linspace(X_p1.min(), X_p1.max(), 300).reshape(-1, 1)
fig, axs = plt.subplots(2, 2, figsize=(11, 8))
for g, ax in zip(range(1, 5), axs.flatten()):
    P_tr = potencias(X_tr, g)
    mu, sigma = P_tr.mean(axis=0), P_tr.std(axis=0)
    W_mc = minimos_cuadrados(matriz_Xa(X_tr, g, mu, sigma), Y_tr)
    ax.scatter(X_p1, Y_p1, s=10, label="Datos")
    ax.plot(x_linea, matriz_Xa(x_linea, g, mu, sigma) @ W_mc, "r-", lw=2, label="Modelo")
    ax.set_title(f"Grado {g}  (J = {costo(matriz_Xa(X_tr, g, mu, sigma), Y_tr, W_mc):.2e})")
    ax.legend(); ax.grid(True)
plt.tight_layout(); plt.show()
````

**Entrenamiento del modelo final (grado óptimo).** Con el grado elegido se prueban varias tasas de aprendizaje, incluyendo una por encima del límite $2/\lambda_{max}$ para comprobar la divergencia.

````python
# ------------------------------------------------------------
# MODELO FINAL: ADALINE CON EL GRADO ÓPTIMO
# ------------------------------------------------------------
P_tr = potencias(X_tr, grado_opt)
mu_opt, sigma_opt = P_tr.mean(axis=0), P_tr.std(axis=0)
Xa_tr = matriz_Xa(X_tr, grado_opt, mu_opt, sigma_opt)
Xa_te = matriz_Xa(X_te, grado_opt, mu_opt, sigma_opt)

eta_lim = eta_maxima(Xa_tr)
print(f"Límite de estabilidad: eta < {eta_lim:.3f}\n")

tasas_p1 = [0.01, 0.1, 0.3, 0.6, round(1.1 * eta_lim, 2)]
epocas_p1 = 50000
historiales, filas = {}, []
for eta in tasas_p1:
    W, J_hist = adaline_gd(Xa_tr, Y_tr, eta, epocas_p1)
    historiales[eta] = J_hist
    J_fin = costo(Xa_tr, Y_tr, W)
    diverge = (not np.isfinite(J_fin)) or J_fin > J_hist[0]
    filas.append({"eta": eta, "Épocas": epocas_p1,
                  "J final (entren.)": np.nan if diverge else J_fin,
                  "J prueba": np.nan if diverge else costo(Xa_te, Y_te, W),
                  "Estado": "Diverge" if diverge else ("Convergió" if J_fin < 1e-10 else "Lento")})
pd.DataFrame(filas)
````

````python
# Convergencia para cada tasa de aprendizaje
plt.figure(figsize=(10, 5))
for eta, J_hist in historiales.items():
    if np.isfinite(J_hist).all() and J_hist[-1] < J_hist[0]:
        plt.loglog(range(1, epocas_p1 + 1), J_hist, label=f"eta = {eta}")
plt.xlabel("Épocas"); plt.ylabel("Costo J")
plt.title(f"Convergencia de ADALINE, grado {grado_opt} (las tasas que divergen se omiten)")
plt.ylim(1e-32, 1e2)
plt.legend(); plt.grid(True, which="both", alpha=0.3)
plt.show()
````

````python
# Modelo final con la mejor tasa de aprendizaje estable
eta_final = 0.6
W_final, J_final_hist = adaline_gd(Xa_tr, Y_tr, eta_final, epocas_p1)
W_mc_opt = minimos_cuadrados(Xa_tr, Y_tr)

# Métricas en el conjunto de prueba
Y_est_te = Xa_te @ W_final
E_te = Y_te - Y_est_te
RMSE_te = np.sqrt(np.mean(E_te**2))
R2_te = 1 - np.sum(E_te**2) / np.sum((Y_te - Y_te.mean())**2)
print(f"RMSE en prueba: {RMSE_te:.3e}")
print(f"R^2 en prueba : {R2_te:.10f}\n")

# Coeficientes del polinomio en la escala original
nombres = ["w0 (1)"] + [f"w{i} (x^{i})" for i in range(1, grado_opt + 1)]
pd.set_option("display.float_format", "{:.6f}".format)
pd.DataFrame({"Mínimos cuadrados": desnormalizar_z(W_mc_opt, mu_opt, sigma_opt),
              "ADALINE (GD)": desnormalizar_z(W_final, mu_opt, sigma_opt)}, index=nombres)
````

````python
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# Curva ajustada sobre datos de entrenamiento y prueba
axs[0].scatter(X_tr, Y_tr, s=12, label="Entrenamiento")
axs[0].scatter(X_te, Y_te, s=25, marker="^", label="Prueba")
axs[0].plot(x_linea, matriz_Xa(x_linea, grado_opt, mu_opt, sigma_opt) @ W_final, "r-", lw=2,
            label=f"ADALINE grado {grado_opt}")
axs[0].set(xlabel="x", ylabel="y", title="Problem 1: modelo final")
axs[0].legend(); axs[0].grid(True)

# Residuos en prueba
axs[1].scatter(Y_est_te, E_te, color="purple", alpha=0.7)
axs[1].axhline(0, color="red", ls="--")
axs[1].set(xlabel="y estimado", ylabel="Residuo (real - estimado)", title="Residuos en el conjunto de prueba")
axs[1].grid(True)

plt.tight_layout(); plt.show()
pd.reset_option("display.float_format")
````

#### 4.3.3 Análisis de resultados

Los valores de $x$ están en $[0, 1]$ y los de $y$ en $[4.04, 7.00]$, sin valores atípicos.

- De grado 2 a grado 3 el costo cae prácticamente a **cero** (del orden de $10^{-29}$, que es solo error de redondeo de la computadora); esa caída equivale al 6 % restante del costo del modelo lineal.
- De grado 3 en adelante la curva queda **plana**: agregar $x^4, x^5, \dots$ ya no mejora nada (las pequeñas variaciones entre $10^{-29}$ y $10^{-21}$ son solo redondeo numérico, que crece con el grado porque la matriz está peor condicionada).

El codo está claramente en **grado 3**, y el criterio automático (mejora menor al 1 % del costo lineal) llega a la misma conclusión.

**3. Validación del grado.** El costo en el conjunto de prueba sigue el mismo patrón que el de entrenamiento, por lo que el grado 3 no es producto de memorizar los datos: el modelo generaliza a puntos que no vio. En la gráfica de los grados 1 a 4 se observa que el grado 1 y 2 se separan de los puntos en los extremos, mientras que los grados 3 y 4 son indistinguibles.

**4. ADALINE frente a mínimos cuadrados en el codo.** Con $\eta = 0.2$ y 20 000 épocas, ADALINE alcanza exactamente el costo de mínimos cuadrados en los grados 1 y 2, y en grado 3 llega a $\approx 10^{-7}$ (con más épocas o una $\eta$ mayor llega a cero, ver la tabla de tasas). Para grados 4 a 8, en cambio, se queda en costos de $10^{-5}$, **unas 100 veces más altos que con grado 3**, aunque en teoría esos modelos pueden ajustar igual o mejor. Esto pasa porque las columnas $x^4, x^5, \dots$ están casi perfectamente correlacionadas con las anteriores, la matriz queda mal condicionada y el gradiente descendente converge muy lento. Es un argumento extra a favor de no usar un grado mayor al necesario: un grado excesivo no solo no ayuda, sino que hace más difícil el entrenamiento.

**5. Tasa de aprendizaje.** Para grado 3 el límite de estabilidad es $\eta < 2/\lambda_{max} \approx 0.69$. En 50 000 épocas, con $\eta = 0.01$ el costo se queda en $\approx 4\times10^{-4}$ y con $\eta = 0.1$ en $\approx 10^{-8}$; con $\eta = 0.3$ baja a $10^{-18}$ y con $\eta = 0.6$ llega al error de máquina ($10^{-29}$). Con $\eta = 0.76$ (10 % arriba del límite) el entrenamiento **diverge**, tal como predice la teoría. Gracias a la normalización z, la convergencia es mucho más rápida que en la Práctica 1 (donde con min-max se necesitaban 100 000 épocas para llegar a $10^{-8}$).

**6. Modelo encontrado.** Al desnormalizar los pesos, ADALINE y mínimos cuadrados coinciden en
$$\hat{y} = 7 + 0\,x - 5\,x^2 + 2\,x^3$$
con RMSE en prueba de $\approx 5\times10^{-15}$ y $R^2 = 1$. Los residuos son del orden de $10^{-14}$; la forma de onda que se ve en su gráfica no es un error del modelo, sino el redondeo de la aritmética de punto flotante. Los coeficientes son números exactos y el término lineal es nulo, lo que indica que los datos fueron **generados con este polinomio y sin ruido**. Por eso el costo llega a cero: en datos reales con ruido el codo se vería igual, pero el costo después del codo se estabilizaría en el nivel del ruido en lugar de llegar a cero, y el costo de prueba podría empezar a subir con grados altos (sobreajuste).

**Conclusión del Problem 1:** el codo de Jambú indica que el **grado óptimo es 3**, y la red ADALINE entrenada por gradiente descendente con normalización z y $\eta = 0.6$ recupera el polinomio $y = 7 - 5x^2 + 2x^3$ con error prácticamente nulo.
